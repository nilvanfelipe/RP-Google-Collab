```python
import matplotlib.pyplot as plt
import pandas as pd

# Lê o arquivo Excel
df = pd.read_excel('/content/Vendas.xlsx')

# Gráfico de barras para mostrar a Receita de Vendas por Canal de Venda
plt.figure(figsize=(10, 6))
df.groupby('Canal Venda')['Receita Vendas'].sum().plot(kind='bar', color='skyblue')
plt.title('Receita de Vendas por Canal de Venda')
plt.xlabel('Canal de Venda')
plt.ylabel('Receita de Vendas')
plt.xticks(rotation=45, ha='right')  # Rotação dos rótulos no eixo x
plt.show()

# Gráfico de barras para mostrar o Lucro por Gênero
plt.figure(figsize=(10, 6))
df.groupby('Gênero')['Lucro'].sum().plot(kind='bar', color='lightcoral')
plt.title('Lucro por Gênero')
plt.xlabel('Gênero')
plt.ylabel('Lucro')
plt.xticks(rotation=0)  # Rotação dos rótulos no eixo x
plt.show()

# Histograma para mostrar a distribuição de Quantidade Pedido
plt.figure(figsize=(10, 6))
plt.hist(df['Qnt Pedido'], bins=20, color='lightgreen', edgecolor='black')
plt.title('Distribuição de Quantidade Pedido')
plt.xlabel('Quantidade Pedido')
plt.ylabel('Frequência')
plt.show()

# Gráfico de pizza para mostrar a distribuição de Forma de Pagamento
plt.figure(figsize=(8, 8))
df['Forma Pagamento'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=['lightblue', 'lightcoral', 'lightgreen'])
plt.title('Distribuição de Forma de Pagamento')
plt.show()
```
