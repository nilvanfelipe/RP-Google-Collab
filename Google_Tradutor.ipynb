## Google-Tradutor:
Este código em Python utiliza a biblioteca googletrans para traduzir um texto fornecido pelo usuário para outro idioma.

```python
pip install googletrans==4.0.0-rc1
```

```python
from googletrans import Translator

def translate_text():
    translator = Translator()

    # Solicitar ao usuário o texto a ser traduzido
    text = input("Digite o texto que você deseja traduzir: ")

    # Solicitar ao usuário o idioma de destino
    target_language = input("Digite o código do idioma de destino (por exemplo, 'en' para inglês): ")

    try:
        # Traduzir o texto
        translation = translator.translate(text, dest=target_language)

        # Exibir a tradução
        print(f"Texto Original: {text}")
        print(f"Tradução ({target_language}): {translation.text}")
    except Exception as e:
        print(f"Ocorreu um erro: {str(e)}")

if __name__ == "__main__":
    print("Bem-vindo ao Tradutor de Texto!")
    translate_text()
```

Aqui estão os principais pontos do código:

- O que essa função faz?
Função translate_text: Ela cria uma instância do tradutor (Translator) da biblioteca googletrans. Solicita ao usuário que insira o texto a ser traduzido. Solicita ao usuário que insira o código do idioma de destino para a tradução. Utiliza a função translate do tradutor para obter a tradução do texto para o idioma de destino. Exibe tanto o texto original quanto a tradução.

Entrada do Usuário: No bloco if name == "main":, exibe uma mensagem de boas-vindas e chama a função translate_text().

Exceções: Inclui um bloco try e except para lidar com exceções que podem ocorrer durante a tradução, como erros de conexão ou problemas com a biblioteca. Em resumo, o código cria uma simples interface de linha de comando para permitir que o usuário insira um texto e escolha um idioma de destino para a tradução, utilizando a biblioteca googletrans para realizar a tradução e exibindo o resultado na tela.
